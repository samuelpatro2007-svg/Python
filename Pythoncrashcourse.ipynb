# Loop Control Structure

```python
# 1. Break
# 2. Continue
# 3. Pass
```

```python
# 1 Break : Breaks the entire loop close to it
for i in range(1,11):
    if i==5 :
        break
    else:
        print(i)
        
```

```python
# 2. Continue : Skips the particular iteration
for i in range(1,11):
    if i==5:
        continue    # Leaves all the code below it and go to the next iteration
    else:
        print(i)
    
```

```python
# 3. Pass : it does nothing
for i in range(1,11):
    pass
```

```python
# Multiple Inheritance
class father:
    def father_fun(self):
        print("I am in fathers class")
        
class mother:
    def mother_fun(self):
        print("I am in mothers class")
        
class son(father,mother):
    def son_fun(self):
        print("I am in sons class")
        
son_obj = son()
son_obj.father_fun()
son_obj.mother_fun()
son_obj.son_fun()
        
```

```python
# Multi-level Inheritance
class Easy:
    def lvl_easy(self):
        print("This level is easy")
        
class Medium(Easy):
    def lvl_medium(self):
        print("This level is medium")
        
class Hard(Medium):
    def lvl_hard(self):
        print("This level is hard")
        
Hard_obj = Hard()
Hard_obj.lvl_hard()
Hard_obj.lvl_medium()
Hard_obj.lvl_easy()
```

```python
# Hierarchy Inheritance
class grandparents:
    def grndprts_fun(self):
        print("This is grand parents house")
        
class parents(grandparents):
    def prts_fun(self):
        print("This is parents room in grandparents house")
        
class children(grandparents):
    def chldrn_fun(self):
        print("This is childrens room in grandparents house")
        
class grandchildren(grandparents):
    def grnd_chldrn_fun(self):
        print("This is Grand childrens room in grandparents house")
        
        
obj_parents = parents()
obj_children = children()
obj_grandchildren = grandchildren()

obj_parents.prts_fun()
obj_parents.grndprts_fun()

obj_children.chldrn_fun()
obj_parents.grndprts_fun()

obj_grandchildren.grnd_chldrn_fun()
obj_parents.grndprts_fun()
```

# Abstraction

```python
from abc import ABC,abstractmethod

class baseClass(ABC):
    @abstractmethod
    def abstract_fun(self):
        pass
    
    def concrete_fun(self):
        print("Base class is calling this concrete function")
        
class hiddenClass(baseClass):        
    def fun(self):
        pass

    def abstract_fun(self):
        print("Hidden class is calling this abstract function")
        
    def concrete_fun(self):
        print("Hidden class is calling this concrete function")
        
hiddenClass_obj = hiddenClass()
hiddenClass_obj.abstract_fun()
hiddenClass_obj.concrete_fun()
```

```python
# Super() functionality
from abc import ABC,abstractmethod

class baseClass(ABC):
    @abstractmethod
    def fun1(self):
        print("It is normal function in baseClass")
        
    def fun2(self):
        pass
    
class hiddenClass(baseClass):
    def fun1(self):
        print("It is a normal function : fun1() in hiddenClass")
        
    def fun2(self):
        print("It is a normal function : fun2() in hiddenClass")
        
    def fun3(self):
        super().fun1()     # This will call your fun1() in base class, but not in current class
    
myobj = hiddenClass()
myobj.fun1()
myobj.fun2()
myobj.fun3()
```

# Encapsulation

```python
class baseClass:
    __var1 = 10   # Private variable
    _var2 = 20    # Protected variable
    var3 = 30     # Public variable
    
    def baseFun(self):
        print("I am in base function")
        print(self.__var1)
        print(self._var2)
        print(self.var3)
        
class derivedClass(baseClass):
    def derivedFun(self):
        print("I am in derived fun")
        # print(self.__var1)
        print(self._var2)
        print(self.var3)
        
        
baseobj = baseClass()
baseobj.baseFun()

derivedobj = derivedClass()
derivedobj.derivedFun()

print("I am in program scope")
# print(baseobj.__var1)
print(baseobj._var2)  # sometimes gives warning
print(baseobj.var3)

         
```

```python
# Task : Bank account
class baseClass():
    __deposit = 500
    _balanceenquiry = 1000 
    withdraw = 500
    
    def baseFun(self):
        print("This is baseClass")
        print("Deposit :",self.__deposit)
        print("Remaining balance :",self._balanceenquiry)        
        print("Withdrawn amount :",self.withdraw)

class derivedClass(baseClass):        
    def derivedFun(self):
        print(" This is derivedClass")
        print(self._balanceenquiry)
        print(self.withdraw)
        
baseobj = baseClass()
baseobj.baseFun()

derivedobj = derivedClass()
derivedobj.derivedFun() 

print("This is program Scope")
# print(baseobj.__deposit)
print(baseobj._balanceenquiry)
print(baseobj.withdraw)
```
